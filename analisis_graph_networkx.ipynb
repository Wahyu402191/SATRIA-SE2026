# Analisis Graph menggunakan NetworkX
## Tugas Pencarian dan Penambangan Web

Notebook ini berisi analisis graph dengan 5 vertex, baik berarah maupun tidak berarah.
Analisis meliputi: degree, rata-rata lintasan terpendek, koefisien klasterisasi, distribusi derajat, rata-rata degree, density, dan adjacency matrix.

## 1. Import Library yang Diperlukan

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter

print("Library berhasil diimport")
print(f"Versi NetworkX: {nx.__version__}")
```

## 2. Membuat Graph Tidak Berarah (Undirected Graph)

```python
# Membuat graph tidak berarah dengan 5 vertex
G_undirected = nx.Graph()

# Menambahkan 5 vertex (node)
vertices = ['A', 'B', 'C', 'D', 'E']
G_undirected.add_nodes_from(vertices)

# Menambahkan edges (hubungan antar vertex)
edges_undirected = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'D'),
    ('C', 'E'),
    ('D', 'E')
]
G_undirected.add_edges_from(edges_undirected)

print("Graph Tidak Berarah berhasil dibuat")
print(f"Jumlah Vertex: {G_undirected.number_of_nodes()}")
print(f"Jumlah Edge: {G_undirected.number_of_edges()}")
```

## 3. Visualisasi Graph Tidak Berarah

```python
# Visualisasi graph tidak berarah
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G_undirected, seed=42)
nx.draw(G_undirected, pos, with_labels=True, node_color='lightblue', 
        node_size=1500, font_size=16, font_weight='bold', 
        edge_color='gray', width=2, arrows=False)
plt.title('Graph Tidak Berarah (Undirected Graph)', fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## 4. Membuat Graph Berarah (Directed Graph)

```python
# Membuat graph berarah dengan 5 vertex yang sama
G_directed = nx.DiGraph()

# Menambahkan 5 vertex (node)
G_directed.add_nodes_from(vertices)

# Menambahkan edges berarah (directed edges)
edges_directed = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'A'),
    ('C', 'D'),
    ('C', 'E'),
    ('D', 'E'),
    ('E', 'D')
]
G_directed.add_edges_from(edges_directed)

print("Graph Berarah berhasil dibuat")
print(f"Jumlah Vertex: {G_directed.number_of_nodes()}")
print(f"Jumlah Edge: {G_directed.number_of_edges()}")
```

## 5. Visualisasi Graph Berarah

```python
# Visualisasi graph berarah
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G_directed, seed=42)
nx.draw(G_directed, pos, with_labels=True, node_color='lightcoral', 
        node_size=1500, font_size=16, font_weight='bold', 
        edge_color='gray', width=2, arrows=True, 
        arrowsize=20, arrowstyle='->', connectionstyle='arc3,rad=0.1')
plt.title('Graph Berarah (Directed Graph)', fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## 6. Menghitung Degree Setiap Vertex - Graph Tidak Berarah

```python
# Degree untuk graph tidak berarah
print("DEGREE SETIAP VERTEX - GRAPH TIDAK BERARAH")
print("="*50)

degree_undirected = dict(G_undirected.degree())

for vertex, degree in sorted(degree_undirected.items()):
    print(f"Vertex {vertex}: {degree}")

print("\nPenjelasan: Degree adalah jumlah edge yang terhubung ke vertex tersebut")
```

## 7. Menghitung Degree Setiap Vertex - Graph Berarah

```python
# Degree untuk graph berarah (in-degree, out-degree, dan total degree)
print("DEGREE SETIAP VERTEX - GRAPH BERARAH")
print("="*70)

in_degree = dict(G_directed.in_degree())
out_degree = dict(G_directed.out_degree())
total_degree = dict(G_directed.degree())

print(f"{'Vertex':<10} {'In-Degree':<15} {'Out-Degree':<15} {'Total Degree':<15}")
print("-"*70)

for vertex in sorted(vertices):
    print(f"{vertex:<10} {in_degree[vertex]:<15} {out_degree[vertex]:<15} {total_degree[vertex]:<15}")

print("\nPenjelasan:")
print("- In-Degree: Jumlah edge yang masuk ke vertex")
print("- Out-Degree: Jumlah edge yang keluar dari vertex")
print("- Total Degree: In-Degree + Out-Degree")
```

## 8. Menghitung Rata-rata Lintasan Terpendek - Graph Tidak Berarah

```python
# Rata-rata lintasan terpendek untuk graph tidak berarah
print("RATA-RATA LINTASAN TERPENDEK - GRAPH TIDAK BERARAH")
print("="*50)

if nx.is_connected(G_undirected):
    avg_shortest_path_undirected = nx.average_shortest_path_length(G_undirected)
    print(f"Rata-rata lintasan terpendek: {avg_shortest_path_undirected:.4f}")
    
    print("\nDetail lintasan terpendek antar semua pasangan vertex:")
    print("-"*50)
    
    shortest_paths = dict(nx.all_pairs_shortest_path_length(G_undirected))
    for source in sorted(vertices):
        for target in sorted(vertices):
            if source != target:
                distance = shortest_paths[source][target]
                print(f"{source} -> {target}: {distance}")
else:
    print("Graph tidak terhubung, tidak dapat menghitung rata-rata lintasan terpendek")
```

## 9. Menghitung Rata-rata Lintasan Terpendek - Graph Berarah

```python
# Rata-rata lintasan terpendek untuk graph berarah
print("RATA-RATA LINTASAN TERPENDEK - GRAPH BERARAH")
print("="*50)

if nx.is_weakly_connected(G_directed):
    try:
        avg_shortest_path_directed = nx.average_shortest_path_length(G_directed)
        print(f"Rata-rata lintasan terpendek: {avg_shortest_path_directed:.4f}")
    except nx.NetworkXError:
        print("Graph berarah tidak strongly connected, menghitung untuk setiap komponen:")
    
    print("\nDetail lintasan terpendek antar semua pasangan vertex:")
    print("-"*50)
    
    for source in sorted(vertices):
        for target in sorted(vertices):
            if source != target:
                try:
                    distance = nx.shortest_path_length(G_directed, source, target)
                    print(f"{source} -> {target}: {distance}")
                except nx.NetworkXNoPath:
                    print(f"{source} -> {target}: Tidak ada jalur")
else:
    print("Graph tidak terhubung")
```

## 10. Menghitung Koefisien Klasterisasi per Vertex - Graph Tidak Berarah

```python
# Koefisien klasterisasi untuk graph tidak berarah
print("KOEFISIEN KLASTERISASI PER VERTEX - GRAPH TIDAK BERARAH")
print("="*50)

clustering_undirected = nx.clustering(G_undirected)

for vertex in sorted(vertices):
    print(f"Vertex {vertex}: {clustering_undirected[vertex]:.4f}")

avg_clustering_undirected = nx.average_clustering(G_undirected)
print(f"\nRata-rata koefisien klasterisasi: {avg_clustering_undirected:.4f}")

print("\nPenjelasan: Koefisien klasterisasi mengukur seberapa cenderung")
print("tetangga-tetangga dari suatu vertex untuk saling terhubung.")
print("Nilai 0 = tidak ada tetangga yang terhubung")
print("Nilai 1 = semua tetangga saling terhubung (membentuk clique)")
```

## 11. Menghitung Koefisien Klasterisasi per Vertex - Graph Berarah

```python
# Koefisien klasterisasi untuk graph berarah
print("KOEFISIEN KLASTERISASI PER VERTEX - GRAPH BERARAH")
print("="*50)

clustering_directed = nx.clustering(G_directed)

for vertex in sorted(vertices):
    print(f"Vertex {vertex}: {clustering_directed[vertex]:.4f}")

avg_clustering_directed = nx.average_clustering(G_directed)
print(f"\nRata-rata koefisien klasterisasi: {avg_clustering_directed:.4f}")

print("\nPenjelasan: Untuk graph berarah, klasterisasi dihitung dengan")
print("mempertimbangkan arah dari setiap edge dalam membentuk segitiga.")
```

## 12. Menghitung Distribusi Derajat - Graph Tidak Berarah

```python
# Distribusi derajat untuk graph tidak berarah
print("DISTRIBUSI DERAJAT - GRAPH TIDAK BERARAH")
print("="*50)

degrees_undirected = [degree for node, degree in G_undirected.degree()]
degree_count_undirected = Counter(degrees_undirected)

print("Degree\tJumlah Vertex")
print("-"*30)
for degree in sorted(degree_count_undirected.keys()):
    print(f"{degree}\t{degree_count_undirected[degree]}")

# Visualisasi distribusi derajat
plt.figure(figsize=(10, 6))
degrees_sorted = sorted(degree_count_undirected.keys())
counts = [degree_count_undirected[d] for d in degrees_sorted]
plt.bar(degrees_sorted, counts, color='lightblue', edgecolor='black')
plt.xlabel('Degree', fontsize=12, fontweight='bold')
plt.ylabel('Jumlah Vertex', fontsize=12, fontweight='bold')
plt.title('Distribusi Derajat - Graph Tidak Berarah', fontsize=14, fontweight='bold')
plt.xticks(degrees_sorted)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 13. Menghitung Distribusi Derajat - Graph Berarah

```python
# Distribusi derajat untuk graph berarah
print("DISTRIBUSI DERAJAT - GRAPH BERARAH")
print("="*50)

in_degrees = [degree for node, degree in G_directed.in_degree()]
out_degrees = [degree for node, degree in G_directed.out_degree()]
total_degrees = [degree for node, degree in G_directed.degree()]

in_degree_count = Counter(in_degrees)
out_degree_count = Counter(out_degrees)
total_degree_count = Counter(total_degrees)

print("\nDistribusi In-Degree:")
print("-"*30)
for degree in sorted(in_degree_count.keys()):
    print(f"In-Degree {degree}: {in_degree_count[degree]} vertex")

print("\nDistribusi Out-Degree:")
print("-"*30)
for degree in sorted(out_degree_count.keys()):
    print(f"Out-Degree {degree}: {out_degree_count[degree]} vertex")

print("\nDistribusi Total Degree:")
print("-"*30)
for degree in sorted(total_degree_count.keys()):
    print(f"Total Degree {degree}: {total_degree_count[degree]} vertex")

# Visualisasi distribusi derajat
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# In-Degree
in_degrees_sorted = sorted(in_degree_count.keys())
in_counts = [in_degree_count[d] for d in in_degrees_sorted]
axes[0].bar(in_degrees_sorted, in_counts, color='lightcoral', edgecolor='black')
axes[0].set_xlabel('In-Degree', fontweight='bold')
axes[0].set_ylabel('Jumlah Vertex', fontweight='bold')
axes[0].set_title('Distribusi In-Degree', fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

# Out-Degree
out_degrees_sorted = sorted(out_degree_count.keys())
out_counts = [out_degree_count[d] for d in out_degrees_sorted]
axes[1].bar(out_degrees_sorted, out_counts, color='lightgreen', edgecolor='black')
axes[1].set_xlabel('Out-Degree', fontweight='bold')
axes[1].set_ylabel('Jumlah Vertex', fontweight='bold')
axes[1].set_title('Distribusi Out-Degree', fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)

# Total Degree
total_degrees_sorted = sorted(total_degree_count.keys())
total_counts = [total_degree_count[d] for d in total_degrees_sorted]
axes[2].bar(total_degrees_sorted, total_counts, color='lightyellow', edgecolor='black')
axes[2].set_xlabel('Total Degree', fontweight='bold')
axes[2].set_ylabel('Jumlah Vertex', fontweight='bold')
axes[2].set_title('Distribusi Total Degree', fontweight='bold')
axes[2].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 14. Menghitung Rata-rata Degree - Graph Tidak Berarah

```python
# Rata-rata degree untuk graph tidak berarah
print("RATA-RATA DEGREE - GRAPH TIDAK BERARAH")
print("="*50)

degrees_list = [degree for node, degree in G_undirected.degree()]
avg_degree_undirected = sum(degrees_list) / len(degrees_list)

print(f"Rata-rata degree: {avg_degree_undirected:.4f}")
print(f"\nDegree minimum: {min(degrees_list)}")
print(f"Degree maksimum: {max(degrees_list)}")
print(f"Total edges: {G_undirected.number_of_edges()}")
print(f"Total vertices: {G_undirected.number_of_nodes()}")

print("\nPenjelasan: Rata-rata degree dihitung dengan menjumlahkan semua")
print("degree vertex dibagi dengan jumlah vertex.")
```

## 15. Menghitung Rata-rata Degree - Graph Berarah

```python
# Rata-rata degree untuk graph berarah
print("RATA-RATA DEGREE - GRAPH BERARAH")
print("="*50)

in_degrees_list = [degree for node, degree in G_directed.in_degree()]
out_degrees_list = [degree for node, degree in G_directed.out_degree()]
total_degrees_list = [degree for node, degree in G_directed.degree()]

avg_in_degree = sum(in_degrees_list) / len(in_degrees_list)
avg_out_degree = sum(out_degrees_list) / len(out_degrees_list)
avg_total_degree = sum(total_degrees_list) / len(total_degrees_list)

print(f"Rata-rata In-Degree: {avg_in_degree:.4f}")
print(f"Rata-rata Out-Degree: {avg_out_degree:.4f}")
print(f"Rata-rata Total Degree: {avg_total_degree:.4f}")

print(f"\nIn-Degree minimum: {min(in_degrees_list)}")
print(f"In-Degree maksimum: {max(in_degrees_list)}")
print(f"Out-Degree minimum: {min(out_degrees_list)}")
print(f"Out-Degree maksimum: {max(out_degrees_list)}")
print(f"\nTotal edges: {G_directed.number_of_edges()}")
print(f"Total vertices: {G_directed.number_of_nodes()}")
```

## 16. Menghitung Density Graph - Graph Tidak Berarah

```python
# Density untuk graph tidak berarah
print("DENSITY GRAPH - GRAPH TIDAK BERARAH")
print("="*50)

density_undirected = nx.density(G_undirected)

print(f"Density: {density_undirected:.4f}")

n = G_undirected.number_of_nodes()
m = G_undirected.number_of_edges()
max_edges = n * (n - 1) / 2

print(f"\nJumlah edges aktual: {m}")
print(f"Jumlah edges maksimum yang mungkin: {int(max_edges)}")
print(f"Persentase edges terisi: {density_undirected * 100:.2f}%")

print("\nPenjelasan: Density mengukur seberapa padat suatu graph.")
print("Nilai 0 = tidak ada edge (graph kosong)")
print("Nilai 1 = semua vertex saling terhubung (complete graph)")
print(f"\nFormula untuk undirected graph: D = 2m / [n(n-1)]")
print(f"Di mana m = jumlah edges, n = jumlah vertices")
```

## 17. Menghitung Density Graph - Graph Berarah

```python
# Density untuk graph berarah
print("DENSITY GRAPH - GRAPH BERARAH")
print("="*50)

density_directed = nx.density(G_directed)

print(f"Density: {density_directed:.4f}")

n = G_directed.number_of_nodes()
m = G_directed.number_of_edges()
max_edges = n * (n - 1)

print(f"\nJumlah edges aktual: {m}")
print(f"Jumlah edges maksimum yang mungkin: {max_edges}")
print(f"Persentase edges terisi: {density_directed * 100:.2f}%")

print("\nPenjelasan: Untuk graph berarah, jumlah edges maksimum adalah")
print("n(n-1) karena setiap pasang vertex bisa punya 2 arah.")
print(f"\nFormula untuk directed graph: D = m / [n(n-1)]")
```

## 18. Adjacency Matrix - Graph Tidak Berarah

```python
# Adjacency matrix untuk graph tidak berarah
print("ADJACENCY MATRIX - GRAPH TIDAK BERARAH")
print("="*50)

adj_matrix_undirected = nx.adjacency_matrix(G_undirected, nodelist=sorted(vertices))
adj_matrix_undirected_dense = adj_matrix_undirected.todense()

print("\nMatrix representasi:")
print("Baris/Kolom: ", " ".join(sorted(vertices)))
print("-"*50)
for i, vertex in enumerate(sorted(vertices)):
    row = adj_matrix_undirected_dense[i].tolist()[0]
    print(f"{vertex}: {row}")

print("\nPenjelasan: Adjacency matrix adalah representasi matriks dari graph.")
print("Nilai 1 menunjukkan ada edge antara dua vertex, 0 menunjukkan tidak ada.")
print("Untuk undirected graph, matrix ini simetris.")
```

## 19. Visualisasi Adjacency Matrix - Graph Tidak Berarah

```python
# Visualisasi adjacency matrix untuk graph tidak berarah
plt.figure(figsize=(8, 6))
plt.imshow(adj_matrix_undirected_dense, cmap='Blues', interpolation='nearest')
plt.colorbar(label='Koneksi')
plt.xticks(range(len(vertices)), sorted(vertices), fontsize=12)
plt.yticks(range(len(vertices)), sorted(vertices), fontsize=12)
plt.xlabel('Vertex', fontsize=12, fontweight='bold')
plt.ylabel('Vertex', fontsize=12, fontweight='bold')
plt.title('Adjacency Matrix - Graph Tidak Berarah', fontsize=14, fontweight='bold')

# Tambahkan nilai di setiap sel
for i in range(len(vertices)):
    for j in range(len(vertices)):
        value = adj_matrix_undirected_dense[i, j]
        plt.text(j, i, str(int(value)), ha='center', va='center', 
                color='white' if value > 0 else 'black', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 20. Adjacency Matrix - Graph Berarah

```python
# Adjacency matrix untuk graph berarah
print("ADJACENCY MATRIX - GRAPH BERARAH")
print("="*50)

adj_matrix_directed = nx.adjacency_matrix(G_directed, nodelist=sorted(vertices))
adj_matrix_directed_dense = adj_matrix_directed.todense()

print("\nMatrix representasi:")
print("     Ke: ", " ".join(sorted(vertices)))
print("Dari:")
print("-"*50)
for i, vertex in enumerate(sorted(vertices)):
    row = adj_matrix_directed_dense[i].tolist()[0]
    print(f"{vertex}: {row}")

print("\nPenjelasan: Untuk directed graph, matrix tidak simetris.")
print("Baris i, kolom j = 1 berarti ada edge dari vertex i ke vertex j.")
print("Matrix[i][j] tidak sama dengan Matrix[j][i] untuk directed graph.")
```

## 21. Visualisasi Adjacency Matrix - Graph Berarah

```python
# Visualisasi adjacency matrix untuk graph berarah
plt.figure(figsize=(8, 6))
plt.imshow(adj_matrix_directed_dense, cmap='Reds', interpolation='nearest')
plt.colorbar(label='Koneksi')
plt.xticks(range(len(vertices)), sorted(vertices), fontsize=12)
plt.yticks(range(len(vertices)), sorted(vertices), fontsize=12)
plt.xlabel('Vertex Tujuan', fontsize=12, fontweight='bold')
plt.ylabel('Vertex Asal', fontsize=12, fontweight='bold')
plt.title('Adjacency Matrix - Graph Berarah', fontsize=14, fontweight='bold')

# Tambahkan nilai di setiap sel
for i in range(len(vertices)):
    for j in range(len(vertices)):
        value = adj_matrix_directed_dense[i, j]
        plt.text(j, i, str(int(value)), ha='center', va='center', 
                color='white' if value > 0 else 'black', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
